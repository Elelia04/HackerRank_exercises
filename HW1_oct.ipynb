```python
#SWAP CASE 

def swap_case(s):
    fin_str = ""
    for el in s:
        if el.isupper():
            fin_str += el.lower()
        elif el.islower():
            fin_str += el.upper()
        else:
            fin_str += el
    return fin_str
```

```python
#TEXT WRAP

def wrap(string, max_width):
    count = 0
    temp_str = ""
    fin_lst = []
    
    for el in string:
        count += 1
        temp_str += el
        
        if count == max_width:
            fin_lst.append(temp_str)
            temp_str = ""
            count = 0
    
    if temp_str:
        fin_lst.append(temp_str)
            
    return "\n".join(fin_lst)
```

```python
#ALPHABET RANGOLI

def print_rangoli(size):
    import string 
    
    alphabet_lst = list(string.ascii_lowercase)
    rangoli_width = 4*size-3
    rows = []
    
    for i in range(size):
        letters = []
        for j in range(size-1, size-1-i-1, -1):
            letters.append(alphabet_lst[j])
            
        mirror = letters[:-1][::-1]
        full_row = letters + mirror
        line = "-".join(full_row)
        centered = line.center(rangoli_width, "-")
        rows.append(centered)
        
    for r in rows + rows[-2::-1]:
        print(r)
```

```python
#TIME DELTA 

def time_delta(t1, t2):
    from datetime import datetime
    
    format_date = "%a %d %b %Y %H:%M:%S %z"
    
    time1 = datetime.strptime(t1, format_date)
    time2 = datetime.strptime(t2, format_date)
    diff = time1 - time2
    seconds = diff.total_seconds()
    return str(int(abs(seconds)))
```

```python
#RE.START(), RE.END()

import re

string = input()
pattern = input()

matches = re.finditer(r'(?=(' + re.escape(pattern) + r'))', string)

positions = [(m.start(1), m.end(1) - 1) for m in matches]

if positions:
    for p in positions:
        print(p)
else:
    print((-1, -1))
```

```python
#HTML PARSER PART 1

from html.parser import HTMLParser

class MyHTMLParser(HTMLParser):
    def handle_starttag(self, tag, attrs):
        print("Start :", tag)
        if attrs:
            for element in attrs:
                attribute_name = element[0]
                attribute_value = element[1]
                print(f"-> {attribute_name} > {attribute_value}")

    def handle_endtag(self, tag):
        print("End   :", tag)

    def handle_startendtag(self, tag, attrs):
        print("Empty :", tag)
        if attrs:
            for element in attrs:
                attribute_name = element[0]
                attribute_value = element[1]
                print(f"-> {attribute_name} > {attribute_value}")


parser = MyHTMLParser()

for i in range(int(input())):
    S = input()
    print(parser.feed(S) if parser.feed(S) else '', end='')

```

```python
#HTML PARSER PART 2

from html.parser import HTMLParser

class MyHTMLParser(HTMLParser):
    def handle_comment(self, data):
        print(">>> Multi-line Comment" if '\n' in data else ">>> Single-line Comment")
        print(data)
        
    def handle_data(self, data):
        if data.strip():
            print(f">>> Data\n{data}")

html = ""       
for i in range(int(input())):
    html += input().rstrip()
    html += '\n'

parser = MyHTMLParser()
parser.feed(html)
parser.close()
```

```python
#DETECT HTML TAGS

from html.parser import HTMLParser

class TagPrinter(HTMLParser):
    def handle_starttag(self, tag, attrs):
        print(tag)
        if attrs:  
            for attr_name, attr_val in attrs:
                print(f"-> {attr_name} > {attr_val}")

lines = int(input())
markup = "\n".join(input() for _ in range(lines))

p = TagPrinter()
p.feed(markup)
```

```python
#VALIDATING UID

import re

n = int(input())

for _ in range(n):
    uid = input().strip()
    if len(uid) != 10:
        print("Invalid")
        continue
    if not uid.isalnum():
        print("Invalid")
        continue
    if len(re.findall(r'[A-Z]', uid)) < 2 or len(re.findall(r'\d', uid)) < 3:
        print("Invalid")
        continue
    if len(set(uid)) != 10:
        print("Invalid")
        continue
    print("Valid")
```

```python
#VALIDATING CREDIT CARD NUMBERS 

import re

n = int(input())
pattern = r'^[4-6]\d{3}(-?\d{4}){3}$'

for _ in range(n):
    card = input().strip()
    if re.fullmatch(pattern, card):
        card = card.replace('-', '')
        if not re.search(r'(\d)\1{3,}', card):
            print("Valid")
        else:
            print("Invalid")
    else:
        print("Invalid")
```

```python
#KANGAROO 

def kangaroo(x1, v1, x2, v2):
    while x1 <= x2 and v1 >v2:
        if x1 == x2:
            return 'YES'
        else:
            x1 += v1
            x2 += v2
    return 'NO'
```

```python
#RECURSIVE DIGIT SUM

def superDigit(n, k):

    n = str(sum([int(i) for i in n])) * k
    def calcSuperDigit(n):
        n = str(sum([int(i) for i in n]))
        if len(n) > 1:
            n = calcSuperDigit(n)
        return n

    return calcSuperDigit(n)
```

```python
#INSERTIONSORT1

def insertionSort1(n, arr):
    e = arr[-1]
    
    for i in range(n-2,-1,-1):
        if arr[i]>e:
            arr[i+1]= arr[i]
            print(*(arr))
        else:
            arr[i+1]= e
            print(*(arr))
            break
    if e not in arr:
        arr[0]=e
        print(*(arr))
```

```python
#INSERTIONSORT2

def insertionSort2(n, arr):
    for i in range(1,n):
        k=i
        while k>0 and arr[k]<arr[k-1]:
            arr[k-1], arr[k] = arr[k], arr[k-1]
            k -= 1
        print(*arr)


```
